```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import classification_report, accuracy_score
```

```python
# Load the dataset
data = pd.read_csv('Milk Data.csv',encoding='windows-1252',lineterminator='\n')

# Data preprocessing
# Convert 'Result' to numerical labels
le = LabelEncoder()
data['Result_encoded'] = le.fit_transform(data['Result\r'])

# Select features and target
features = ['Fat', 'SNF', 'Temperature°C']  # Using these 3 features for 3D visualization
X = data[features]
y = data['Result_encoded']
```

```python
# Split data into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Standardize features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python

# Train SVM model
svm_model = SVC(kernel='rbf', C=1.0, gamma='scale', probability=True)
svm_model.fit(X_train_scaled, y_train)
```

```python

# Predictions
y_pred = svm_model.predict(X_test_scaled)
```

```python
# Evaluate model
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=le.classes_))
```

```python
# 3D Visualization
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')
# Create a mesh grid for the decision boundary
x_min, x_max = X_train_scaled[:, 0].min() - 1, X_train_scaled[:, 0].max() + 1
y_min, y_max = X_train_scaled[:, 1].min() - 1, X_train_scaled[:, 1].max() + 1
z_min, z_max = X_train_scaled[:, 2].min() - 1, X_train_scaled[:, 2].max() + 1

xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.1),
                     np.arange(y_min, y_max, 0.1))

# Plot decision boundary for one class at a time
for i, class_name in enumerate(le.classes_):
    # Get the probability of the current class
    Z = svm_model.predict_proba(np.c_[xx.ravel(), yy.ravel(), 
                                np.zeros_like(xx.ravel()) * (z_max - z_min) + z_min])[:, i]
    Z = Z.reshape(xx.shape)
        # Plot the surface
    surf = ax.plot_surface(xx, yy, Z, cmap=plt.cm.coolwarm, alpha=0.3,
                          linewidth=0, antialiased=False)
    
    # Plot the actual data points
    mask = (y_train == i)
    ax.scatter(X_train_scaled[mask, 0], X_train_scaled[mask, 1], X_train_scaled[mask, 2],
               label=f'{class_name}', s=20, alpha=0.7)
    # Add labels and title
ax.set_xlabel('Fat (Standardized)')
ax.set_ylabel('SNF (Standardized)')
ax.set_zlabel('Temperature (Standardized)')
ax.set_title('3D Visualization of Milk Adulteration Detection using SVM')

# Add legend
ax.legend()

plt.tight_layout()
plt.show()
```
